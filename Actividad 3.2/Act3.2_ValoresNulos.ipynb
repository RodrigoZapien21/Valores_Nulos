```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('barcelona.csv')
data.head(5)
```

```python
data.info()
```

```python
columnas_links= [
    "listing_url", "picture_url", "host_url", "host_profile_url",
    "host_picture_url", "description", "host_about"
]
 
columnas_100_pct_nulas = [
    "neighborhood_overview", "host_since", "host_response_time",
    "host_response_rate", "host_acceptance_rate", "host_thumbnail_url",
    "host_neighbourhood", "host_total_listings_count", "host_verifications",
    "neighbourhood", "calendar_updated", "instant_bookable"
]
 
columnas_redundantes = [
    "scrape_id", "last_scraped", "calendar_last_scraped", "source",
    "host_profile_id", "hosts_time_as_user_months", "hosts_time_as_host_months",
    "minimum_minimum_nights", "maximum_minimum_nights",
    "minimum_maximum_nights", "bathrooms_text",
    "price_quote_checkin_date", "price_quote_checkout_date",
    "price_quote_total_price", "price_quote_raw", "price_quote_price_per_night",
    "availability_eoy", "has_availability",
    "calculated_host_listings_count_entire_homes",
    "calculated_host_listings_count_private_rooms",
    "calculated_host_listings_count_shared_rooms",
]

columnas_a_eliminar = (columnas_links + columnas_100_pct_nulas + columnas_redundantes)
```

```python
df = data.drop(columns=columnas_a_eliminar)
print("Columnas después de la limpieza inicial:", df.shape[1])
```

```python
df["price"] = df["price"].str.replace("$", "", regex=False)
df["price"] = df["price"].str.replace(",", "", regex=False)
df["price"] = df["price"].astype(float)
```

```python
valores_nulos = df.isnull().sum()
valores_nulos
```

```python
df["host_name"] = df["host_name"].fillna("Desconocido")
df["host_is_superhost"] = df["host_is_superhost"].fillna(df["host_is_superhost"].mode()[0])
df["host_has_profile_pic"] = df["host_has_profile_pic"].fillna(df["host_has_profile_pic"].mode()[0])
df["host_identity_verified"] = df["host_identity_verified"].fillna(df["host_identity_verified"].mode()[0])
df["hosts_time_as_user_years"] = df["hosts_time_as_user_years"].fillna(df["hosts_time_as_user_years"].median())
df["hosts_time_as_host_years"] = df["hosts_time_as_host_years"].fillna(df["hosts_time_as_host_years"].median())
df["host_listings_count"] = df["host_listings_count"].fillna(df["host_listings_count"].median())
```

```python
df["host_location"] = df["host_location"].fillna("No especificado")
df["bathrooms"] = df["bathrooms"].fillna(df["bathrooms"].median())
df["bedrooms"] = df["bedrooms"].fillna(df["bedrooms"].median())
df["beds"] = df["beds"].fillna(df["beds"].median())
df["price"] = df["price"].fillna(df["price"].median())
df["estimated_revenue_l365d"] = np.where(
    df["estimated_revenue_l365d"].isnull(),
    df["price"] * df["estimated_occupancy_l365d"],
    df["estimated_revenue_l365d"]
)

df["first_review"] = df["first_review"].fillna("Sin reseñas")
df["last_review"] = df["last_review"].fillna("Sin reseñas")
```

```python
#Se consideran las columnas scores para poder llenarlas con los valores de mediana en todo los calculos de cada una de las columnas
#ya que si se asignara en escala de 0-5, sería incorrecto y reflejaria un mal servicio cuando en realidad es una variable sin informacion
columnas_scores = [
    "review_scores_rating", "review_scores_accuracy", "review_scores_cleanliness",
    "review_scores_checkin", "review_scores_communication",
    "review_scores_location", "review_scores_value"
]
for columna in columnas_scores:
    df[columna] = df[columna].fillna(df[columna].median())
```

```python
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
df["license"] = df["license"].fillna("Sin licencia registrada")
```

```python
#Se consideran aqullas columnas que casi no tienen valores nulos (menos de 5)
#y se aplica una mediana general de estas columnas que se basan en minimum y maximum nights
columnas_numericas_restantes = df.select_dtypes(include=[np.number]).columns
for columna in columnas_numericas_restantes:
    if df[columna].isnull().sum() > 0:
        df[columna] = df[columna].fillna(df[columna].median())
```
